```python
def events_in_window(events, meta):
    if "target" in meta.columns:
        ev = events.merge(meta[['cookie_id', 'window_start_ts', 'window_end_ts', "target"]], on='cookie_id')
    else:
        ev = events.merge(meta[['cookie_id', 'window_start_ts', 'window_end_ts']], on='cookie_id')
    return ev[(ev.event_ts >= ev.window_start_ts) & (ev.event_ts < ev.window_end_ts)]

ev_tr = events_in_window(events, train)
ev_te = events_in_window(events, test)
print(len(ev_tr), len(ev_te))
print(ev_tr["cookie_id"].nunique())
ev_tr["search_query"].nunique()
```

```python
import pandas as pd

# 1. Приводим user_agent к нижнему регистру
ev_tr['user_agent_lower'] = ev_tr['user_agent'].str.lower()

# 2. Список всех маркеров (в нижнем регистре)
bot_patterns = [
    'headlesschrome',
    'phantomjs',
    'selenium',
    'puppeteer',
    'python-requests',
    'curl',
    'wget',
    'bot',
    'crawler',
    'spider'
]

# 3. Общая маска — есть ли хотя бы один маркер
pattern_regex = '|'.join(bot_patterns)
mask_any = ev_tr['user_agent_lower'].str.contains(pattern_regex, na=False, regex=True)

# 4. Подсчёт уникальных cookie_id по target для общей маски
print("=== Любой маркер бота ===")
print(ev_tr[mask_any].groupby('target')['cookie_id'].nunique())
print()

# 5. Детализация по каждому маркеру отдельно
results = []
for pattern in bot_patterns:
    mask = ev_tr['user_agent_lower'].str.contains(pattern, na=False, regex=False)
    subset = ev_tr[mask]
    n_0 = subset[subset['target'] == 0]['cookie_id'].nunique()
    n_1 = subset[subset['target'] == 1]['cookie_id'].nunique()
    results.append({
        'pattern': pattern,
        'target_0': n_0,
        'target_1': n_1,
        'total': n_0 + n_1
    })

df_results = pd.DataFrame(results).sort_values('total', ascending=False)

print("=== Детализация по маркерам ===")
print(df_results.to_string(index=False))
print()

# 6. Дополнительно: сколько строк (не уникальных cookie_id) с маркерами
print("=== Строк с маркерами по target ===")
print(ev_tr[mask_any]['target'].value_counts())
print()

# 7. Примеры подозрительных UA
print("=== Примеры UA с маркерами ===")
print(ev_tr[mask_any][['cookie_id', 'user_agent', 'target']].head(10))
```

```python
import numpy as np
import pandas as pd

# Работаем с копией, чтобы не менять исходный ev_tr
ev = ev_tr.copy()

# Приводим event_ts к datetime и сортируем
ev['event_ts'] = pd.to_datetime(ev['event_ts'])
ev = ev.sort_values(['cookie_id', 'event_ts'])

# Считаем интервалы между событиями
ev['dt'] = ev.groupby('cookie_id')['event_ts'].diff().dt.total_seconds()

# Агрегируем признаки по cookie_id
time_features = ev.groupby('cookie_id').agg(
    dt_mean=('dt', 'mean'),
    dt_median=('dt', 'median'),
    dt_std=('dt', 'std'),
    dt_min=('dt', 'min'),
    dt_max=('dt', 'max'),
    dt_q25=('dt', lambda x: x.quantile(0.25)),
    dt_q75=('dt', lambda x: x.quantile(0.75)),
    target=('target', 'first')
).reset_index()

# Коэффициент вариации
time_features['dt_cv'] = time_features['dt_std'] / (time_features['dt_mean'] + 1e-6)

# Доля самого частого (округлённого) интервала
def mode_rate(x):
    x = x.dropna().round()
    if len(x) == 0:
        return np.nan
    return x.value_counts(normalize=True).iloc[0]

dt_mode = ev.groupby('cookie_id')['dt'].apply(mode_rate).reset_index(name='dt_mode_rate')

time_features = time_features.merge(dt_mode, on='cookie_id', how='left')

# Округляем признаки
cols = ['dt_mean', 'dt_median', 'dt_std', 'dt_min', 'dt_max', 'dt_q25', 'dt_q75', 'dt_cv', 'dt_mode_rate']
time_features[cols] = time_features[cols].round(2)

time_features
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

features = [
    'dt_mean',
    'dt_median',
    'dt_std',
    'dt_min',
    'dt_max',
    'dt_q25',
    'dt_q75',
    'dt_cv',
    'dt_mode_rate'
]

for feature in features:
    plt.figure(figsize=(10, 5))
    sns.histplot(data=time_features, x=feature, hue='target', bins=50, kde=True, stat='density', common_norm=False, alpha=0.4)
    plt.title(f'Распределение {feature}')
    plt.xlabel(feature)
    plt.ylabel('Количество cookie_id')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

features = [
    'dt_mean',
    'dt_median',
    'dt_std',
    'dt_min',
    'dt_max',
    'dt_cv',
]

# Ограничиваем выбросы, иначе pairplot будет нечитаемым
plot_df = time_features[features + ['target']].copy()

# Обрезаем по 1-му и 99-му процентилю для каждого признака
for col in features:
    lo, hi = plot_df[col].quantile([0.01, 0.99])
    plot_df[col] = plot_df[col].clip(lo, hi)

sns.pairplot(
    plot_df,
    hue='target',
    diag_kind='kde',        # на диагонали — плотность
    corner=True,            # только нижний треугольник (компактнее)
    plot_kws={'alpha': 0.4, 's': 15},
    diag_kws={'fill': True, 'alpha': 0.4},
)
plt.suptitle('Pairplot временных признаков по target', y=1.02)
plt.show()
```
