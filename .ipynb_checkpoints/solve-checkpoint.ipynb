```python
import pandas as pd
import seaborn as sns
import matplotlib as plt
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from catboost import CatBoostClassifier
```

```python
train = pd.read_csv('data/train.csv', parse_dates=['cookie_created_at', 'window_start_ts', 'window_end_ts'])
test = pd.read_csv('data/test.csv', parse_dates=['cookie_created_at', 'window_start_ts', 'window_end_ts'])
events = pd.read_csv('data/events.csv', parse_dates=['event_ts'])
```

## Удаление одинаковых строк в events

```python
print('Размер до удаления:', events.shape)
print('Полных дубликатов:', events.duplicated().sum())
events = events.drop_duplicates().reset_index(drop=True)
print('Размер после удаления:', events.shape)
```

## Оставляет только те cookie_id, которые попали во временное окно и формируем DataFrame для обучения и теста 

```python
def events_in_window(events, meta):
    if "target" in meta.columns:
        ev = events.merge(meta[['cookie_id', 'window_start_ts', 'window_end_ts', "target"]], on='cookie_id')
    else:
        ev = events.merge(meta[['cookie_id', 'window_start_ts', 'window_end_ts']], on='cookie_id')
    return ev[(ev.event_ts >= ev.window_start_ts) & (ev.event_ts < ev.window_end_ts)]

ev_tr = events_in_window(events, train)
ev_te = events_in_window(events, test)
print(len(ev_tr), len(ev_te))
```

# Признаки по времени

У меня опявилась гипотиза, что время, которое тратит бот на то или иное действие должно быть меньше, чем время настоящего человека 

```python
# Работаем с копией, чтобы не менять исходный ev_tr
ev = ev_tr.copy()

# Приводим event_ts к datetime и сортируем
ev['event_ts'] = pd.to_datetime(ev['event_ts'])
ev = ev.sort_values(['cookie_id', 'event_ts'])

# Считаем интервалы между событиями
ev['dt'] = ev.groupby('cookie_id')['event_ts'].diff().dt.total_seconds()

# Агрегируем признаки по cookie_id
time_features = ev.groupby('cookie_id').agg(
    dt_mean=('dt', 'mean'),
    dt_median=('dt', 'median'),
    dt_std=('dt', 'std'),
    dt_min=('dt', 'min'),
    dt_q25=('dt', lambda x: x.quantile(0.25)),
    dt_q75=('dt', lambda x: x.quantile(0.75)),
    target=('target', 'first')
).reset_index()

# Коэффициент вариации
time_features['dt_cv'] = time_features['dt_std'] / (time_features['dt_mean'] + 1e-6)

# Доля самого частого (округлённого) интервала
def mode_rate(x):
    x = x.dropna().round()
    if len(x) == 0:
        return np.nan
    return x.value_counts(normalize=True).iloc[0]

dt_mode = ev.groupby('cookie_id')['dt'].apply(mode_rate).reset_index(name='dt_mode_rate')

time_features = time_features.merge(dt_mode, on='cookie_id', how='left')

# Округляем признаки
cols = ['dt_mean', 'dt_median', 'dt_std', 'dt_min', 'dt_q25', 'dt_q75', 'dt_cv', 'dt_mode_rate']
time_features[cols] = time_features[cols].round(2)

time_features.head()
```

```python
for feature in time_features.drop(columns=['cookie_id', "target"]).columns:
    plt.figure(figsize=(10, 5))
    sns.histplot(data=time_features, x=feature, hue='target', bins=50, kde=True, stat='density', common_norm=False, alpha=0.4)
    plt.title(f'Распределение {feature}')
    plt.xlabel(feature)
    plt.ylabel('Количество cookie_id')
    plt.tight_layout()
    plt.show()
```

## Сравнение платформ, которыми пользуются боты и люди

```python
print(ev_tr["platform"].value_counts())
# Так как все платформы написаны в разных регистрах, необходимо приввести все значения к одному регистру
ev_tr["platform"] = ev_tr["platform"].str.lower()
print("\n")
print(ev_tr["platform"].value_counts())
```

```python
sns.histplot(data=ev_tr, x='platform', hue='target',
             multiple='dodge', stat='probability', common_norm=False) 
```

## На графике видно, что боты больше пользуются web и desktop, а люди android и ios. Можно сделать вывод, что это будет хороший признак для нашей задачи

```python
platform_counts = (
    ev_tr.groupby(['cookie_id', 'platform'])
    .size()
    .unstack(fill_value=0)
    .add_prefix('cnt_platform_')
)
platform_counts

platform_flags = (
    (platform_counts > 0)
    .astype(int)
    .add_prefix('is_platform_')
)

platform_share = (
    platform_counts
    .div(platform_counts.sum(axis=1), axis=0)
    .add_prefix('share_platform_')
)


cnt_raw = ev_tr.groupby(['cookie_id', 'platform']).size().unstack(fill_value=0)

# Сначала main_platform — на «чистых» колонках
summary = pd.DataFrame(index=cnt_raw.index)
summary['n_platforms'] = (cnt_raw > 0).sum(axis=1)
summary['main_platform'] = cnt_raw.idxmax(axis=1)          # ← 'web', 'android', ...
summary['main_platform_share'] = cnt_raw.max(axis=1) / cnt_raw.sum(axis=1)

summary
```

## Обработка user_agent 

```python
import pandas as pd

# Работаем с копией
ev = ev_tr.copy()

# 1. Приводим user_agent к нижнему регистру (временно)
ua_lower = ev['user_agent'].str.lower()

# 2. Список маркеров
bot_patterns = [
    'headlesschrome',
    'phantomjs',
    'selenium',
    'puppeteer',
    'python-requests',
    'curl',
    'wget',
    'bot',
    'crawler',
    'spider'
]

# 3. Регулярка с границами слов для bot/crawler/spider,
#    чтобы не ловить 'robot', 'botanical', 'abbot' и т.п.
pattern_regex = (
    r'headlesschrome|phantomjs|selenium|puppeteer|python-requests|curl|wget'
    r'|\b(bot|crawler|spider)\b'
)

# 4. Общая маска
mask_any = ua_lower.str.contains(pattern_regex, na=False, regex=True)

# 5. Флаг на уровне cookie_id:
#    strange_word = 1, если хотя бы одно событие этой куки содержит маркер
strange_word_df = (
    ev.assign(strange_word=mask_any.astype(int))
      .groupby('cookie_id', as_index=False)['strange_word']
      .max()
)

print(strange_word_df.head())
print('Всего cookie_id:', len(strange_word_df))
print('С флагом 1:', (strange_word_df['strange_word'] == 1).sum())
print('С флагом 0:', (strange_word_df['strange_word'] == 0).sum())
```

# Также значение std показывает, что боты в большей степени смотрят одинаковое количество категорий, в отличие от людей -> количество уникальных категорий может помочь в детекции ботов. (признак дал прирост 0.04 к целевой метрике)

```python
# Формируем DataFrame: для каждой куки — количество уникальных категорий
cat_diversity = (
    ev_tr.groupby('cookie_id')['item_category']
    .nunique()
    .rename('cat_nunique')
    .reset_index()
)

targets = ev_tr.groupby('cookie_id')['target'].first().reset_index()
cat_diversity = cat_diversity.merge(targets, on='cookie_id', how='left')
cat_diversity.head()
```

## Признакм по event_name

```python
import pandas as pd

ev = ev_tr.copy()

# Бинарные флаги: 1, если событие встречается у куки хотя бы раз
event_flags = (
    ev.assign(flag=1)
    .pivot_table(
        index='cookie_id',
        columns='event_name',
        values='flag',
        aggfunc='max',       # max → 1, если было хоть одно событие
        fill_value=0,
    )
    .add_prefix('has_')      # has_item_view, has_login, ...
    .reset_index()
)

# Присоединяем target
targets = ev.groupby('cookie_id')['target'].first().reset_index()
event_flags = event_flags.merge(targets, on='cookie_id', how='left')

print(event_flags.head())
print()
print('Колонки:', event_flags.columns.tolist())
```

## Признаки по location

```python
import pandas as pd

# Работаем с копией
ev = ev_tr.copy()

# Уникальные item_location для каждой куки
loc_features = (
    ev.groupby('cookie_id')['item_location']
    .nunique()
    .rename('item_location_nunique')
    .reset_index()
)

# Присоединяем target
targets = ev.groupby('cookie_id')['target'].first().reset_index()
loc_features = loc_features.merge(targets, on='cookie_id', how='left')

# Сводка по target
print(
    loc_features
    .groupby('target')['item_location_nunique']
    .agg(['mean', 'median', 'std', 'count'])
    .round(3)
)

print()
print(loc_features.head())
```

## Объединение всех признаков в один DataFrame 

```python
def merge_features(df_with_feature, feature):
    if "target" in feature.columns:
        cols = feature.drop(columns=["target"]).columns
    else:
        cols = feature.columns
        
    return df_with_feature.merge(
        feature[cols],
        on='cookie_id',
        how='left'
    )

def basic_features(ev, meta):
    g = ev.groupby('cookie_id')
    f = pd.DataFrame({
        'n_events': g.size(),
        'item_nunique': g['item_id'].nunique(),
    }).reset_index()

    #Присоединяем к meta — все cookie_id из meta сохраняются
    result = meta.merge(f, on='cookie_id', how='left')
    return result

df_with_feature = train[['cookie_id', "target"]]
df_with_feature = merge_features(df_with_feature, time_features)

df_with_feature = merge_features(df_with_feature, platform_counts)
df_with_feature = merge_features(df_with_feature, platform_flags)
df_with_feature = merge_features(df_with_feature, platform_share)
df_with_feature = merge_features(df_with_feature, summary)

df_with_feature = merge_features(df_with_feature, strange_word_df)

df_with_feature = merge_features(df_with_feature, cat_diversity)

df_with_feature = merge_features(df_with_feature, search_diversity)
df_with_feature = merge_features(df_with_feature, event_flags)

df_with_feature = merge_features(df_with_feature, loc_features)

df_with_feature = basic_features(ev_tr, df_with_feature)


df_with_feature
```

```python
print(train.index.equals(df_with_feature.index))
```

```python
is_valid = train.window_start_ts.ge('2026-04-17').values
x = df_with_feature.drop(columns=["target", "cookie_id"])
y = df_with_feature["target"]
X_train, y_train = x[~is_valid], y[~is_valid]
X_valid, y_valid = x[is_valid],  y[is_valid]
```

## Обучение модели 

```python
model = CatBoostClassifier(
    #cat_features=["main_platform"],
    iterations=3000,
    random_strength=5,
    auto_class_weights='Balanced',
    eval_metric='PRAUC',
    random_state=0,
    verbose=100,
    early_stopping_rounds=500,
    depth=3,                    
    l2_leaf_reg=8,
    learning_rate=0.03,
)

model.fit(
    X_train, y_train,
    eval_set=(X_valid, y_valid),
    use_best_model=True,
)
```

```python
from metric import precision_at_recall


predict = model.predict_proba(X_valid)[:, 1]
print('P@R0.7 на валидации:', round(precision_at_recall(y_valid, predict), 4))
```

```python
from catboost import CatBoostClassifier
from metric import precision_at_recall
import pandas as pd
import time

# Сетка параметров
depths = [3, 4, 5, 6]
l2_regs = [3, 5, 8, 12]
learning_rates = [0.02, 0.03, 0.05]
random_strengths = [1, 3, 5]

results = []
total = len(depths) * len(l2_regs) * len(learning_rates) * len(random_strengths)
i = 0

for depth in depths:
    for l2 in l2_regs:
        for lr in learning_rates:
            for rs in random_strengths:
                i += 1
                t0 = time.time()

                model = CatBoostClassifier(
                    iterations=3000,
                    depth=depth,
                    l2_leaf_reg=l2,
                    learning_rate=lr,
                    random_strength=rs,
                    auto_class_weights='Balanced',
                    eval_metric='PRAUC',
                    random_state=0,
                    verbose=0,
                    early_stopping_rounds=100,
                )

                model.fit(
                    X_train, y_train,
                    eval_set=(X_valid, y_valid),
                    use_best_model=True,
                )

                proba = model.predict_proba(X_valid)[:, 1]
                score = precision_at_recall(y_valid, proba)
                elapsed = time.time() - t0

                results.append({
                    'depth': depth,
                    'l2_leaf_reg': l2,
                    'learning_rate': lr,
                    'random_strength': rs,
                    'P@R0.7': round(score, 4),
                    'best_iter': model.get_best_iteration(),
                    'time_sec': round(elapsed, 1),
                })

                print(f'[{i}/{total}] depth={depth}, l2={l2}, lr={lr}, rs={rs} '
                      f'-> P@R0.7={score:.4f} (iter={model.get_best_iteration()}, {elapsed:.0f}s)')

# Итоговая таблица, отсортированная по метрике
df_results = pd.DataFrame(results).sort_values('P@R0.7', ascending=False).reset_index(drop=True)

print()
print('=== Топ-10 комбинаций ===')
print(df_results.head(10).to_string(index=False))

# Лучшая комбинация
best = df_results.iloc[0]
print()
print('=== Лучшая комбинация ===')
print(f"depth         = {best['depth']}")
print(f"l2_leaf_reg   = {best['l2_leaf_reg']}")
print(f"learning_rate = {best['learning_rate']}")
print(f"random_strength = {best['random_strength']}")
print(f"P@R0.7        = {best['P@R0.7']}")
print(f"best_iter     = {best['best_iter']}")
```
